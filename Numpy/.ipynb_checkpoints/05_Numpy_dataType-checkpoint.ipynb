**1.Introduction to Numpy Data types**

Numpy arrays are homomgenous,meaning that they can only store elements of the same type.THis is different from Phython lists, which can hold mixed data types. Numpy supports varios datatyoes(also called dtypes),and understanding them is crusial for optimizing and performance.

**common Data types in Numpy:**
- int32,int64:intezer types with different bit sizes.
- float32,float64: Floating-point types with different precision.
- bool: Boolean datatype.
- complex64, complex128: Complex Number types.
- object: For storing objects (e.g., Python,strings).
- objects:For storing objects (e.g.,Python objects,strings).

```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 4, 5])
print(arr.dtype)  # Output: int64 (or int32 depending on the system)
```

**2.Chaniging Data Types**

you can cast the data type of n array using the .astype() method . this is useful when you need to change the type for a specific operation or when you want to reduce memory usage.

```python
arr = np.array([1.5,2.7,3.9])
print(arr.dtype)
```

```python
arr_int = arr.astype(np.int32)
print(arr_int)
```

```python
print(arr_int.dtype)
```

Example: Downcasting to save Memory

```python
arr_large = np.array([100000, 200000, 300000], dtype=np.int64)
```

```python
arr_small = arr_large.astype(np.int32)
```

```python
print(arr_small)
```

```python
print(arr_small.dtype)
```

**3.Why Datatypes Matter in Numpy**
- Memory Usage: Smaller datatypes use less memory.
- performance: Operatioin on smaller datatype are fater due  to less datatype
                  being processed.
- precision: choosing the appropriate datatype ensures that you don;t lose presicion (e.g., using float32 instead of float64 if you don't need the extra precision).

```python
#Exammple: Memory Usage
arr_int64 = np.array([1,2,3], dtype=np.int64)
arr_int32 = np.array([1,2,3], dtype=np.int32)
print(arr_int64.nbytes)#3 element * 8 bytes each
print(arr_int32.nbytes)#12 bytes(3 elements * 4 bytes each)
```

```python
#string DAtatype in Numpy
#dtype='str' or dtype='U'
arr = np.array(['apple','banana','cherry'], dtype='U10')#unicode string array
print(arr)
```

```python
#Complex Numbers
arr = np.array([1+2j,3+4j,5+6j], dtype='complex64')
print(arr)
```

```python
print(arr.dtype)
```

```python
#Object Datatype
arr = np.array([{'a':1},[1,2,3],'hello'], dtype=object)
print(arr)
```
