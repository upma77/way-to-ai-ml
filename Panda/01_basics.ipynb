**DataFrame - 2D labeled Table**

a data frame is like a dictionary od series-multiple columns with labels.

```python
data = {
"name":["Alice","BOB","Charlie"],
"Age": [25,36,45],
"City":["patna","dehradun","banras"]}
df = pd.DataFrame(data)
df.index = ["a", "b", "c"]
df.columns = ["Name", "Age", "City"]
print(df)
```

**Index and Labels**
Every series and dataframe has an index - it helps with:
-Fast lookups
-Aligning data
-Merging and joining
-Time series operations
